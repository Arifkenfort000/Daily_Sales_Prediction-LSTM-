```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
```

```python
sales = np.array(
    [100, 110, 120, 115, 130, 140, 135, 150,
     160, 155, 170, 180, 175, 190, 200, 195],
    dtype=np.float32
)

X=[]
Y=[]

for i in range(len(sales) - 3):
    X.append(sales[i:i+3])
    Y.append(sales[i+3])


X=np.array(X)
Y=np.array(Y)

# print(np.array(X))
# print(np.array(Y))

X = X.reshape(X.shape[0],X.shape[1],1)

model = Sequential([
    LSTM(16,input_shape= (3,1)),
    Dense(1)
])

model.compile(optimizer='adam',loss='mse')

model.fit(
    X,
    Y,
    epochs=100,
    verbose=1
)
```

```python

last_three = np.array([175, 190, 200], dtype=np.float32)
last_three = last_three.reshape(1, 3, 1)

prediction = model.predict(last_three, verbose=0)

print("Predicted next sales:", prediction[0][0])
```
